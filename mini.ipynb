```python
def create_prompt(ticket_id, ticket_text):

    prompt = f"""
You are a customer support analyst.

Analyze the support ticket and return ONLY a valid JSON object.

JSON Format:

{{
  "ticket_id": "{ticket_id}",
  "category": "",
  "priority": "",
  "sentiment": "",
  "summary": "",
  "recommended_team": ""
}}

Possible categories:
- Login Issue
- Payment Issue
- Technical Issue
- Delivery Issue
- Account Issue
- Other

Possible priorities:
- Low
- Medium
- High
- Critical

Possible sentiments:
- Positive
- Neutral
- Negative

Ticket:
{ticket_text}

Return JSON only.
"""

    return prompt
```

```python
sample_ticket = """
I paid for my order using UPI.
Money was deducted from my account.
Order was not created.
"""

prompt = create_prompt(101, sample_ticket)

print(prompt)
```

```python
response = generator(
    prompt,
    max_new_tokens=256,
    do_sample=False,
    temperature=0.1
)

output = response[0]["generated_text"]

print(output)
```

```python
import re

def extract_json(text):

    match = re.search(r'\{.*\}', text, re.DOTALL)

    if match:
        return match.group()

    return None
```

```python
json_text = extract_json(output)

print(json_text)
```

```python
import json

ticket_data = json.loads(json_text)

print(ticket_data)
```
